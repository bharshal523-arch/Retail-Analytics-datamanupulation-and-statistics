```python
df.groupby(['City'])['Revenue'].sum()
```

```python
category_revenue = pd.pivot_table(
    df,
    values="Revenue",
    index="Category",
    aggfunc=np.sum
)
category_revenue
```

```python
City_revenue = pd.pivot_table(
    df,
    values="Revenue",
    index="City",
    aggfunc=np.sum
)
City_revenue
```

```python
segment_revenue = pd.pivot_table(
    df,
    values="Revenue",
    index="Customer_Segment",
    aggfunc="sum"
)

segment_revenue
```

```python
category_summary = pd.pivot_table(
    df,
    values="Revenue",
    index="Category",
    aggfunc=["sum", "mean", "max", "min"]
)

category_summary
```

```python
channel_summary = pd.pivot_table(
    df,
    values="Units_Sold",
    index="Sales_Channel",
    aggfunc=["sum", "mean"]
)

channel_summary
```

```python
city_performance = pd.pivot_table (
    df,
    values=["Revenue", "Profit_Margin"],
    index="City",
    aggfunc={
        "Revenue": "sum",
        "Profit_Margin": "mean"
    }

)
city_performance
```

```python
city_performance = pd.pivot_table (
    df,
    values=["Revenue", "Profit_Margin", "Brand_Loyalty", "Category"],
    index = "City",
    aggfunc={
        "Revenue": "sum",
        "Profit_Margin": "mean",
        "Brand_Loyalty": "mean",
        "Category": "count"
       
            
  }            
)
city_performance
```

```python
store_summary = pd.pivot_table(
    df,
    values=["Revenue", "Units_Sold"],
    index="Store_Type",
    aggfunc="sum"
)

store_summary
```
